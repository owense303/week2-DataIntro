## 1. Research Question

- **Main research question:** 
Can self-reported screen time and sleep quality responses predict whether adolescents’ BDI-II score crosses the clinical threshold for depression?

- **Why is this question interesting or important?** Due to the extended time spent online and prevalence of depression in adolescents (4.7% in 10-19 year olds globally/1.3% aged 10-14 years), it is important to evaluate factors related to depression in this cohort.

- **Who might use or care about the answer?** Stakeholders concerned with this topic include researchers, government, mental health professionals/organisations, schools, parents and adolescents.

- **Possible sub-questions:** Are screen time and sleep related to any specific items on the  BDI-II?

### 1. Where did the data come from?

### 2. Why was it originally collected?

**Answer:** The data was collected to evaluate the proposed links between screen time, sleep, and depression in adolescents, specifically adolescents in Sweden, aged 12-16 years old.

### 3. Who is represented?

**Answer:** 
Population: Adolescents aged 12-16 years old
Groups: Boys (50.85%) and Girls
Location: Stockholm, Sweden
Time Period: 30th August 2016 - 22nd November 2019 (~ 3.25 years)
Sampling Frame: 55 elementary schools in Stockholm

### 4. Who might be missing?

**Answer:** 
Purposefully excluded groups: 
- Those outside of the desired age cohort (12-16 years old) (justification: group was not of interest for this study)
- Those who did not identify as either male or female (justification: low number of non-binary people within this population)
- Those who with missing responses in the gender and age category
- Those who reported suicidal plans/attempts within the past 2 weeks (justification: were instead offered emergency suicide intervention)
- Those who were exposed to previous psychoeducation/suicidal interventions (justification: as they aimed to alter the mental health trajectory)


Excluded groups not listed:
- People who resided outside of Stockholm

### 5. What personal or sensitive information does it contain, explicitly or implicitly?

**Answer:** 
No names: Data is pseudonymized. It does contain 21 individual BDI-II item responses (clinical depression screening items), sleep-quality items, and self-reported gender which are all from minors. 
Implicit/proxy risk: Combining gender + age + specific BDI-II item patterns is a much richer fingerprint than typical anonymized data; the paper itself flags this by removing mice imputation identifiers specifically for pseudonymization purposes before public release.

### 6. What could your analysis reveal that was not obvious in the original data?

**Answer:** 
The original study used a bifactor Structural Equation Model to derive a latent "G-factor" depression score. Meaning it is not a simple sum. If we instead take a plain sum of the 21 BDI-II items (0–63 range, clinical cutoff >13 for "Mild Depression," as the paper states), the model may recover cruder patterns than the original SEM captured, or could mislabel severity if we not careful about this scoring distinction.

Government: may guide policies/recommendations related to screen time in this cohort.

Mental health professionals/organisations: may be used to guide treatments/group interventions as well as public information and psychoeducation.

### 8. Who could be harmed or misrepresented?

**Answer:** 
Non-binary individuals:
- Participants were excluded because there were too few of them in the sample to analyze reliably as their own group.

Girls:
- Could be unfairly stereotyped if people oversimplify the finding into "screen time causes depression in girls," ignoring that poor sleep is actually the main driver, not screen time directly.

Boys:
- Could be wrongly reassured that "screen time is fine for boys," when really the study just didn't measure the kind of harm (behavioral/attention problems) that may affect them instead.

At-risk youth:
- Adolescents with suicidal thoughts were removed from the data, so any conclusions about "how depressed Swedish teens are" leave out the most vulnerable group, making the problem look smaller than it really is.
